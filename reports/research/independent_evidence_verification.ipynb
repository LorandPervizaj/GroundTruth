# Independent evidence verification
Frozen private inputs; no DB/network access. Source agreement is not publisher truth. See docs/INDEPENDENT_EVIDENCE_VERIFICATION.md for interpretation.

```python
from pathlib import Path
import json
from collections import Counter
from groundtruth.analytics.evidence_verification import matching_metrics, field_metrics
root = Path.cwd()
if root.name == 'research': root = root.parent.parent
evidence = json.loads((root / 'reports/generated/independent_evidence_v1/frozen_evidence_private.json').read_text(encoding='utf8'))
print('Snapshots:', len(evidence['snapshot_reviews']), 'fields:', len(evidence['field_reviews']))
```

```python
result = matching_metrics(evidence['matching_reviews'])
print(json.dumps(result, indent=2))
```

```python
metrics = field_metrics(evidence['field_reviews'])
for source, strata in metrics.items():
    price = strata['random_segment_stratum']['price']
    print(source, 'price agreement', price['counts'].get('source_agreement', 0), '/', price['comparable'], 'reference 95%', price['wilson_reference_95'])
print('Dates:', dict(Counter(r['publication_representation'] for r in evidence['snapshot_reviews'])))
```
